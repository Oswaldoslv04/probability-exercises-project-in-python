## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva una cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
import numpy as np
```

```python
# 1) Fijamos una semilla para que el experimento sea reproducible
np.random.seed(42)

# 2) Definimos cuántas veces repetiremos el experimento (simulación)
n_iter = 1000
```

```python
# 3) Inicializamos contador de casos favorables
conteo_favorables = 0
```

```python
# 4) Ejecutamos el experimento n_iter veces
for i in range(n_iter):
    
    # Lanzamos los dos dados (uno a la vez)
    dado1 = np.random.randint(1,7)
    dado2 = np.random.randint(1,7)
    
    # Calculamos la suma obtenida
    suma = dado1 + dado2
    
    # Verificamos si se cumple el evento:
    # "la suma es mayor a 7 O es un número par"
    if suma > 7 or suma % 2 == 0:
        conteo_favorables += 1
```

```python
# 5) Calculamos probabilidad estimada
prob_estimada = conteo_favorables / n_iter

# 6) Mostramos resultados
print("Total simulaciones:", n_iter)
print("Casos favorables:", conteo_favorables)
print("Probabilidad estimada:", prob_estimada)
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Tomas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
import numpy as np
```

```python
ball_box = {}

# Crea la cajas con las bolas
for i in range(60):
    if i < 10:
        ball_box[i] = "White"
    elif (i > 9) and (i < 30):
        ball_box[i] = "Red"
    else:
        ball_box[i] = "Green"

print(ball_box)
```

```python
# Número de simulaciones
n_iter = 1000
```

```python
# Inicializamos contadores
conteo_3W_2R = 0
conteo_mismo_color = 0
```

```python
# Sacamos 5 bolas (con reemplazo)
for _ in range(n_iter):
    
    # Sacar 5 bolas con reemplazo
    muestra = []
    for _ in range(5):
        # Generamos índice aleatorio entre 0 y 59
        idx = np.random.randint(0, 60)
        # Buscamos el color en ball_box
        muestra.append(ball_box[idx])

    
    # Contamos colores
    blancas = muestra.count("White")
    rojas = muestra.count("Red")
    verdes = muestra.count("Green")
    
        # Evento 1: tomamos 3 blancas y 2 rojas
    if blancas == 3 and rojas == 2:
        conteo_3W_2R += 1
    
    # Evento 2: tomamos todas del mismo color
    if blancas == 5 or rojas == 5 or verdes == 5:
        conteo_mismo_color += 1
```

```python
# Probabilidades estimadas
prob_3W_2R = conteo_3W_2R / n_iter
prob_mismo_color = conteo_mismo_color / n_iter

# Imprimos los resultados.
print(f"De {n_iter} simulaciones:")
print("Probabilidad estimada de 3 blancas y 2 rojas:", prob_3W_2R)
print(f"- {conteo_3W_2R} vez obtuvimos 3 blancas y 2 rojas.")
print("Probabilidad estimada de todas del mismo color:", prob_mismo_color)
print(f"- {conteo_mismo_color} veces salieron todas del mismo color.")
```
